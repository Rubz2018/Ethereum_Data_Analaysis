```python
from google.cloud import bigquery
import pandas as pd
import os
from google.cloud import bigquery
import pandas_gbq
from tqdm import tqdm
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = "/Users/dr.rubaiyatislam/Documents/big_query/keyfile.json"
client = bigquery.Client()
```

```python
query3 = """ 
SELECT 
  address,
  is_erc20,
  is_erc721,
  DATE(block_timestamp) AS block_timestamp,
  block_number,

FROM
  bigquery-public-data.crypto_ethereum.contracts as contracts
WHERE
  #address = '0xc0a47dFe034B400B47bDaD5FecDa2621de6c4d95'
  DATE(block_timestamp) >= '2022-08-18' AND DATE(block_timestamp) <= '2023-08-18'
  #DATE(block_timestamp) >= '2022-09-15' AND DATE(block_timestamp) <= '2023-03-01'
  
ORDER BY block_timestamp
"""
```

```python
df2 = pandas_gbq.read_gbq(
    query3, project_id=client.project, progress_bar_type='tqdm')

df2.head()
```
