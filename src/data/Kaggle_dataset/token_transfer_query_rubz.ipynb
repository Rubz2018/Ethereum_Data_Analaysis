```python
from google.cloud import bigquery
import pandas as pd
import os
from google.cloud import bigquery
import pandas_gbq
from tqdm import tqdm
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = "/Users/dr.rubaiyatislam/Documents/big_query/keyfile.json"
client = bigquery.Client()
```

```python
query1 = """ 
SELECT 
  token_address,
  from_address,
  to_address,
  value,
  transaction_hash,
  DATE(block_timestamp) AS block_timestamp,
  block_number,
  block_hash

FROM
  bigquery-public-data.crypto_ethereum.token_transfers as token_transfers
WHERE
   DATE(block_timestamp) > '2022-11-06' AND DATE(block_timestamp) <= '2022-11-15'
   #block_timestamp > timestamp_sub(current_timestamp, INTERVAL 1 DAY)
ORDER BY block_timestamp
"""
```

```python
df1 = pandas_gbq.read_gbq(
    query1, project_id=client.project, progress_bar_type='tqdm')

df1.head()
```

```python
# Exporting the dataframe to a text file:
df1.to_csv("/Users/dr.rubaiyatislam/Documents/big_query/token_transfer_Pos_2022_11_06_11_15.txt",sep=',')
```
