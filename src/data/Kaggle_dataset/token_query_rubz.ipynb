```python
from google.cloud import bigquery
import pandas as pd
import os
from google.cloud import bigquery
import pandas_gbq
from tqdm import tqdm
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = "/Users/dr.rubaiyatislam/Documents/big_query/keyfile.json"
client = bigquery.Client()
```

```python
query5 = """ 
SELECT 
  address,
  name,
  decimals,
  total_supply,
  DATE(block_timestamp) AS block_timestamp,
  block_number,
  block_hash

FROM
  bigquery-public-data.crypto_ethereum.tokens as tokens
WHERE
  DATE(block_timestamp) >= '2022-09-15' AND DATE(block_timestamp) <= '2023-03-01'
  #block_timestamp > timestamp_sub(current_timestamp, INTERVAL 1 DAY)
ORDER BY block_timestamp
"""
```

```python
df = pandas_gbq.read_gbq(
    query5, project_id=client.project, progress_bar_type='tqdm')
```
