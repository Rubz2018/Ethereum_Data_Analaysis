# Block table and the attributes from 2015 to 2018
- The following query explores all the attributes of the block

- The following query works if we omit 'hash' attribute. 
- We need to fix the hash attribute error issue. 

Test Query

```python
# This works on my pc and on kaggle
query2 = """ 
SELECT 
  DATE(timestamp) AS timestamp,
  number AS block_number,
  blocks.hash AS hsh,
  parent_hash,
  nonce,
  miner,
  difficulty,
  size,
  gas_limit,
  gas_used,
  transaction_count
  base_fee_per_gas
  
FROM
  bigquery-public-data.crypto_ethereum.blocks AS blocks
WHERE
  DATE(timestamp) >= '2022-09-15' AND DATE(timestamp) <= '2023-03-01'
ORDER BY timestamp
"""
```

```python

# Execute the query and convert the results to a Pandas DataFrame
df2=  pandas_gbq.read_gbq(
    query2, project_id=client.project, progress_bar_type='tqdm')
df2.head()
```
