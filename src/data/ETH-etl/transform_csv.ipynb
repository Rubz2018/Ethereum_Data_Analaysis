# Data collection process

[Ethereum etl](https://ethereum-etl.readthedocs.io/en/latest/) was used to collect data. Please read the docs for more info

### Steps
1. pip3 install ethereum-etl
2. input the following command in CLI - 
   ***ethereumetl export_blocks_and_transactions --start-block 16540917 --end-block 16540941 --provider-uri https://mainnet.infura.io/v3/7aef3f0cd1f64408b163814b22cc643c --blocks-output blocks.csv --transactions-output transactions.csv***
3. *blocks.csv* and *transactions.csv* is created containing all informations of specified blocks
4. Data was transformed using Pandas library

```python
import time
from ethereumetl.providers.auto import get_provider_from_uri
from ethereumetl.exporters import CsvItemExporter

start_block = 0
end_block = 5000
provider_uri = 'https://mainnet.infura.io/v3/7aef3f0cd1f64408b163814b22cc643c'
blocks_output = 'blocks.csv'
transactions_output = 'transactions.csv'

start_time = time.time()

provider = get_provider_from_uri(provider_uri)

block_fields = [
    'number', 'hash', 'parent_hash', 'nonce', 'sha3_uncles', 'logs_bloom', 'transactions_root',
    'state_root', 'receipts_root', 'miner', 'difficulty', 'total_difficulty', 'size', 'extra_data',
    'gas_limit', 'gas_used', 'timestamp', 'transaction_count'
]
transaction_fields = [
    'block_number', 'hash', 'nonce', 'block_hash', 'transaction_index', 'from_address',
    'to_address', 'value', 'gas', 'gas_price', 'input'
]

block_exporter = CsvItemExporter(open(blocks_output, 'w'), fields_to_export=block_fields)
transaction_exporter = CsvItemExporter(open(transactions_output, 'w'), fields_to_export=transaction_fields)

for block_number in range(start_block, end_block + 1):
    block = provider.make_request('eth_getBlockByNumber', [hex(block_number), True])
    response = provider.make_request('eth_getBlockTransactionCountByNumber', [hex(block_number)])
    print(response)
    transactions = int(str(response['result']), 16)
    for transaction_index in range(transactions):
        transaction = provider.make_request('eth_getTransactionByBlockNumberAndIndex', [hex(block_number), hex(transaction_index)])
        transaction['block_number'] = block_number
        transaction_exporter.export_item(transaction)
    block_exporter.export_item(block)

block_exporter.close()
transaction_exporter.close()

end_time = time.time()
execution_time = end_time - start_time
print('Execution time:', execution_time)


end_time = time.time()
execution_time = end_time - start_time
print('Execution time:', execution_time)

```
